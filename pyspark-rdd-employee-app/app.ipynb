```python
from pyspark import SparkContext
```

```python
sc = SparkContext("local[*]", "EmployeeRDD")
```

```python
rdd = sc.textFile("data/employees.csv")
header = rdd.first()
```

```python
employees = (
    rdd.filter(lambda row: row != header)
       .map(lambda row: row.split(","))
       .map(lambda row: (
           int(row[0]),      # id
           row[1],           # name
           row[2],           # department
           int(row[3])       # salary
       ))
)
```

```python
sorted_emp = employees.sortBy(
    lambda emp: emp[3],
    ascending=False
)

print("Employees Sorted By Salary")

for emp in sorted_emp.collect():
    print(emp)

sorted_emp.map(
    lambda emp: f"{emp[0]},{emp[1]},{emp[2]},{emp[3]}"
).saveAsTextFile("output/sorted_employees")
```

```python
dept_salary = (
    employees
    .map(lambda emp: (emp[2], emp[3]))
    .reduceByKey(lambda x, y: x + y)
)

print("\nDepartment Wise Salary Totals")

for row in dept_salary.collect():
    print(row)

dept_salary.map(
    lambda x: f"{x[0]},{x[1]}"
).saveAsTextFile("output/department_salary")
```

```python
top3 = sorted_emp.take(3)

print("\nTop 3 Highest Paid Employees")

for emp in top3:
    print(emp)

sc.parallelize(
    [f"{emp[0]},{emp[1]},{emp[2]},{emp[3]}" for emp in top3]
).saveAsTextFile("output/top3_highest_paid")
sc.stop()
```
