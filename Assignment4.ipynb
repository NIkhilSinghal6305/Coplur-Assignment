```python
# 1. Write a Python function to find the maximum of three numbers.

def maximun(a,b,c):
    if(a>b & a>c):
        return a
    elif(b>a & b>c):
        return b
    else:
        return c
    
max = maximun(2,9,8)
print("Maximum : ", max)
```

```python
# 2. Write a Python function that takes a list and returns a new list with distinct elements from the first list.

def distinct_elem(list):
    new_list = []

    for x in list:
        if x not in new_list:
            new_list.append(x)
    return new_list

lst = [1,2,2,3,4,4,5,4]
new_lst = distinct_elem(lst)
print("List with distinct elements : ", new_lst)
```

```python
# 3. Write a Python function to multiply all the numbers in a list.

def mul_elem(lst):
    result=1
    for i in lst:
        result = result * i
    return result

answer = mul_elem([1,2,3,4,5])
print(answer)
```

```python
# 4. Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument.

def factorial(n):
    if n<0:
        return "Not possible"
    
    factorial = 1

    for i in range(1,n+1):
        factorial *= i
    return factorial

n = int(input("Entar a number"))
fact = factorial(n)
print("Factorial : ", fact)
```

```python
# 5. Write a Python program to reverse a string.

def reverse_string(str):
    rev=""
    for i in range(len(str)-1,-1,-1):
        rev += str[i]
    return rev

str = "Nikhil"
rev_str = reverse_string(str)
print(rev_str)
```

```python
# 6. Write a Python function to check whether a number falls within a given range.

def range(n,rstart,rstop):
    if rstart <= n <=rstop:
        return "Number falls whithin range"
    else:
        return "Number doesn't falls whithin range"
    
n = int(input("Enter a number"))
result = range(n,200,877)
print(result)
```

```python
# 7. Write a Python function to Print Even Numbers from a Given List

def even_list(lst):
    elst = []
    for x in lst:
        if x%2==0:
            elst.append(x)
    return elst

lst = [1,2,3,4,5,6,7,8,9,0]
even_list = even_list(lst)
print(even_list)
```

```python
def prime(n):
    if n <= 1:
        return 0

    for x in range(2, n,1):
        if n % x == 0:
            return 0

    return 1


n = int(input("Enter: "))

answer = prime(n)

if answer == 1:
    print("Prime")
else:
    print("Not prime")
```

```python

# 9. Write a Python function that accepts a string and counts the number of upper and lower case letters. 

def count(str):

    lower = 0
    upper = 0
    for x in str:
        if 'a' <= x <= 'z':
            lower+=1
        elif 'A' <= x <= 'Z':
            upper+=1
        else:
            continue
    
    return [lower , upper]

str = "Nikhil SiNGhal"
ans = count(str)
lower = ans[0]
upper = ans[1]
print(lower, upper)
```
