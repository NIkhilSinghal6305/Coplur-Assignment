```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

arr = np.nan_to_num(arr, nan=0)
print(arr)

print(arr.T)

arr3d = np.arange(24).reshape(2, 3, 4)

print(arr3d.shape)

moved = np.moveaxis(arr3d, 0, 2)

print(moved.shape)
print(moved)

arr = np.array([[10, np.nan, 30],
                [40, 50, np.nan],
                [70, 80, 90]])

col_mean = np.nanmean(arr, axis=0)

inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])

print(arr)

arr = np.array([10, -5, 20, -8, 15, -2])

arr[arr < 0] = 0

print(arr)

arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]])

arr2 = np.array([[7, 8, 9],
                 [4, 5, 6]])

avg = (arr1 + arr2) / 2

combined = np.concatenate((arr1.flatten(), arr2.flatten()))

mean = np.mean(combined)
median = np.median(combined)

values, counts = np.unique(combined, return_counts=True)
mode = values[np.argmax(counts)]

print(avg)
print(mean)
print(median)
print(mode)

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])

solution = np.linalg.solve(A, B)

print(solution)

A_inv = np.linalg.inv(A)

result = np.dot(A_inv, B)

print(result)

subjects = ["Sub1", "Sub2", "Sub3", "Sub4", "Sub5"]

sem1 = [8.1, 7.5, 8.0, 7.8, 8.4]
sem2 = [8.5, 8.0, 8.2, 8.1, 8.7]

plt.figure(figsize=(8, 5))

plt.plot(subjects, sem1, marker='o', label="Semester 1")
plt.plot(subjects, sem2, marker='o', label="Semester 2")

plt.title("Semester Result Comparison")
plt.xlabel("Subjects")
plt.ylabel("SGPA / Grade Points")
plt.legend()
plt.grid(True)

plt.show()
```
