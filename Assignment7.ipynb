```python
import pandas as pd

# 1) DataFrame using Two-Dimensional Python List
data1 = [
    [101, "Nikhil", 85],
    [102, "Mohit", 90],
    [103, "Rahul", 78]
]

df1 = pd.DataFrame(data1)

print("1) DataFrame using Two-Dimensional List")
print(df1)
print()


# 2) DataFrame from Python Dictionary
data2 = {
    "ID": [101, 102, 103],
    "Name": ["Nikhil", "Mohit", "Rahul"],
    "Marks": [85, 90, 78]
}

df2 = pd.DataFrame(data2)

print("2) DataFrame from Dictionary")
print(df2)
print()


# 3) DataFrame using List of Lists
data3 = [
    [101, "Nikhil", 85],
    [102, "Mohit", 90],
    [103, "Rahul", 78]
]

df3 = pd.DataFrame(data3, columns=["ID", "Name", "Marks"])

print("3) DataFrame using List of Lists")
print(df3)
print()


# 4) DataFrame using List of Tuples
data4 = [
    (101, "Nikhil", 85),
    (102, "Mohit", 90),
    (103, "Rahul", 78)
]

df4 = pd.DataFrame(data4, columns=["ID", "Name", "Marks"])

print("4) DataFrame using List of Tuples")
print(df4)
print()


# 5) DataFrame from List of Dictionaries
data5 = [
    {"ID": 101, "Name": "Nikhil", "Marks": 85},
    {"ID": 102, "Name": "Mohit", "Marks": 90},
    {"ID": 103, "Name": "Rahul", "Marks": 78}
]

df5 = pd.DataFrame(data5)

print("5) DataFrame from List of Dictionaries")
print(df5)
```

```python
import pandas as pd

# Create DataFrame
data = {
    "ID": [101, 102, 103, 104],
    "Name": ["Nikhil", "Mohit", "Rahul", "Aman"],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)
print()


# 1) Different ways to iterate over rows

print("Using iterrows()")
for index, row in df.iterrows():
    print(index, row["Name"], row["Marks"])

print()

print("Using itertuples()")
for row in df.itertuples():
    print(row.ID, row.Name, row.Marks)

print()


# 2) Selecting rows based on condition

print("Rows where Marks > 80")
print(df[df["Marks"] > 80])
print()


# 3) Select any row using iloc[]

print("Select row at index 2 using iloc[]")
print(df.iloc[2])
print()


# 4) Limited rows selection with given columns

print("First 2 rows with Name and Marks columns")
print(df.loc[0:1, ["Name", "Marks"]])
print()


# 5) Drop rows based on condition

print("Drop rows where Marks < 80")
df2 = df[df["Marks"] >= 80]

print(df2)
print()


# 6) Insert row at given position

new_row = {
    "ID": 105,
    "Name": "Karan",
    "Marks": 88
}

top = df.iloc[:2]
bottom = df.iloc[2:]

df3 = pd.concat([top, pd.DataFrame([new_row]), bottom]).reset_index(drop=True)

print("DataFrame after inserting row at position 2")
print(df3)
print()


# 7) Create list from rows in DataFrame

row_list = df.values.tolist()

print("List created from DataFrame rows")
print(row_list)
```
